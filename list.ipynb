```python
marks=[70,60,80,79,90]
print("Enter the original list: ",marks)
marks.insert(1,95)
print("inserted element:",95)
print("deleted element:",marks.pop(0))
marks[1]=90
print("updated element at index 1 to:",90)
print("updated list:",marks)
print("sorted list:",sorted(marks))
print("searched elements at index:",marks.index(90))
```
