```python
using positive, negative indexing,slicing,occurence,replace in string
```

```python
s="Advanced Python"
print("---positive indexing---")
print("5 th position:",s[5])
print("inbetween 2 and 7:",s[2:7])

print("---negative indexing---")
print("5 th position negative",s[-5])
print("inbetween -7 and -2",s[-7:-2])

print("---slicing---")
print("before 5 :",s[:5])#before 5
print("after 5:",s[5:])#after 5
print("inbetween4 & 10 ",s[4:10])
print("inbetween negative -10 & -4:",s[-10:-4])

print("---ocuurance---")
print("occurance of a :",s.count("a"))
print("occurance of n :",s.count("n"))

print("---replace---")
print("replace:d",s.replace("python","language"))
```

```python
s=input("Enter a String:")
d=input("Enter string 2:")
print("Change the String into upper case:",s.upper())
print("change the string into lower case:",s.lower())
print("Strip the string:",s.strip())
print("splitted string:",s.split())
print("Titled string:",s.title())
print("Capitalize the string:",s.capitalize())
print("joined string :", " and ".join([s, d]))
```

```python
decimal to binary,decimal to octal,decimal to hexadecimal in number system
```

```python
n=int(input("Enter a decimal number inbetween 10-16:"))
print("Decimal to binary:",bin(n)[2:])
print("Decimal to octal:",oct(n)[2:])
print("Decimal to hexadecimal:",hex(n)[2:].upper())
```

```python
insertion,deletion,update,searching,sorting in list of student mark
```

```python
marks=[70,60,80,79,90]
print("Enter the original list: ",marks)
marks.insert(1,95)
print("inserted element:",95)
print("deleted element:",marks.pop(0))
marks[1]=90
print("updated element at index 1 to:",90)
print("updated list:",marks)
print("sorted list:",sorted(marks))
print("searched elements at index:",marks.index(90))
```

```python
numbers = []
n = int(input("Enter the number of elements: "))
for i in range(n):
    numbers.append(int(input("Enter a number: ")))

numbers.sort()

print("Second Largest Number:", numbers[-2])
```

to count even and odd numbers in list

```python
numbers = [10, 15, 20, 25, 30, 35]
even = 0
odd = 0

for number in numbers:
    if number % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```
